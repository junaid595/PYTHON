```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#load the csv dataset
df = pd.read_csv(r"C:\Users\Admin\Documents\data.csv")
print("Dataset :")
print(df)

#line plot
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()

#bar plot
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()

#scatter plot
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()

#correlation matrix and heatmap
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("correlation Matrix :")
print(correlation)
plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("correlation Heatmap")
plt.show()

#pie chart
y = df["Python"]

plt.pie(y, labels = df["Student"])
plt.title("Python Marks of Students")
plt.show()

#histogram
plt.figure(figsize=(8, 6))
plt.hist(df['Maths'], bins=3, edgecolor='black')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.title('Histogram of Maths Marks')
plt.show()
```
